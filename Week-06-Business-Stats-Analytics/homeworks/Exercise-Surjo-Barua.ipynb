## Dataset Setup

First, let's generate the synthetic dataset you'll be analyzing:

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set random seed for reproducible results
np.random.seed(42)

# Generate 500 orders from 2023
n_orders = 500

# Create order IDs and customer data
order_id = np.arange(1, n_orders + 1)
customer_id = np.random.randint(1, 101, size=n_orders)
customer_segment = np.random.choice(["Student", "Professional", "Senior"], size=n_orders, p=[0.4, 0.4, 0.2])
region = np.random.choice(["North", "South", "East", "West"], size=n_orders)
order_date = pd.date_range(start="2023-01-01", periods=n_orders, freq="D")
product_category = np.random.choice(["Electronics", "Clothing", "Books", "Home", "Sports"], size=n_orders)
quantity = np.random.randint(1, 5, size=n_orders)
unit_price = np.round(np.random.uniform(10, 200, size=n_orders), 2)
revenue = quantity * unit_price
discount_applied = np.random.choice([0, 5, 10, 15, 20, 25, 30], size=n_orders, p=[0.5,0.1,0.1,0.1,0.05,0.05,0.1])
payment_method = np.random.choice(["Credit Card", "PayPal", "Gift Card", "Debit Card"], size=n_orders)

# Build dataframe
df = pd.DataFrame({
    "order_id": order_id,
    "customer_id": customer_id,
    "customer_segment": customer_segment,
    "region": region,
    "order_date": order_date,
    "product_category": product_category,
    "quantity": quantity,
    "unit_price": unit_price,
    "revenue": revenue,
    "discount_applied": discount_applied,
    "payment_method": payment_method
})

# Save dataset for reference
df.to_csv("ecommerce_data.csv", index=False)
print("✅ Dataset created successfully!")

# Display first few rows
df.head()
```

## Business Questions

Now answer each of the following business questions. Use the starter code above as a reference, and expand your analysis to solve all questions.

### 1. Revenue Overview
What was the company's total revenue last quarter? Break it down by month.

```python
# Your code here
import pandas as pd
data = pd.read_csv("ecommerce_data.csv")

## convert the order dates to date time
data['order_date'] = pd.to_datetime(data['order_date'])
##Filter the last quater 
start_date = '2023-01-01'
end_date = '2025-09-30'
mask = (data['order_date'] >= start_date) & (data['order_date'] <= end_date)
quarter_data = data[mask]

#Group by month & sum revenue
quarter_data['month'] = quarter_data['order_date'].dt.to_period('M')
monthly_revenue = quarter_data.groupby('month')['revenue'].sum()
print(monthly_revenue)
#calculate the total quater revenue 
total_quarter_revenue = monthly_revenue.sum()
print('Total revenue in last quarter:', total_quarter_revenue)


```

### 2. Top Customers
Who are the top 5 customers by total revenue contribution?

```python
# Your code here
# Calculate revenue per customer and sort
top_customers = data.groupby('customer_id')['revenue'].sum().sort_values(ascending=False).head(5)
print('Top 5 customers by total revenue:')
print(top_customers)

```

### 3. Segment Analysis
Which customer segment (Student, Professional, Senior) spends the most on average per order?

```python
# Your code here
# Average spend per order for each segment
segment_avg = data.groupby('customer_segment')['revenue'].mean().sort_values(ascending=False)
print("Average order value per segment:")
print(segment_avg)

# Segment with highest average
top_segment = segment_avg.idxmax()
print("Segment that spends most on average per order:", top_segment)

```

### 4. Regional Trends
Which region generated the highest revenue? Which region had the highest average order size?

```python
# Your code here
# Region with highest revenue in whole data
region_revenue = data.groupby('region')['revenue'].sum().sort_values(ascending=False)
top_revenue_region = region_revenue.idxmax()
print("Region with highest total revenue:", top_revenue_region)

# Region with highest average order size
region_avg_order = data.groupby('region')['revenue'].mean().sort_values(ascending=False)
top_avg_order_region = region_avg_order.idxmax()
print("Region with highest average order size:", top_avg_order_region)
```

### 5. Product Category Performance
Which product category is most popular by quantity vs. by revenue?

```python
# Your code here
# Most popular category by quantity
category_quantity = data.groupby('product_category')['quantity'].sum().sort_values(ascending=False)
print("Category by total quantity sold:")
print(category_quantity)

# Most popular category by revenue
category_revenue = data.groupby('product_category')['revenue'].sum().sort_values(ascending=False)
print("Category by total revenue generated:")
print(category_revenue)
```

### 6. Discount Effectiveness
Do discounted orders generate higher or lower revenue per order compared to non-discounted orders?

```python
# Your code here
# Create flag for discounted orders
discounted = data['discount_applied'] > 0

# Average revenue per order for discounted and non-discounted
avg_rev_discounted = data[discounted]['revenue'].mean()
avg_rev_nondiscounted = data[~discounted]['revenue'].mean()
print("Average revenue per discounted order:", avg_rev_discounted)
print("Average revenue per non-discounted order:", avg_rev_nondiscounted)
```

### 7. Payment Method Usage
What percentage of orders use each payment method? Does any payment method correlate with higher spending?

```python
# Your code here
# Percentage of orders by payment method
payment_counts = data['payment_method'].value_counts(normalize=True) * 100
print("Percent of orders by payment method:")
print(payment_counts)

# Average revenue by payment method
payment_avg_revenue = data.groupby('payment_method')['revenue'].mean().sort_values(ascending=False)
print("Average revenue by payment method:")
print(payment_avg_revenue)
```

### 8. Seasonality Check
Plot revenue by day of week — are weekends busier than weekdays?

```python
# Your code here
# Add day of week column
data['day_of_week'] = pd.to_datetime(data['order_date']).dt.day_name()

# Revenue by day of week
day_revenue = data.groupby('day_of_week')['revenue'].sum().sort_values(ascending=False)
print("Total revenue by day of week:")
print(day_revenue)

# Plot revenue by day of week
import matplotlib.pyplot as plt
day_revenue = day_revenue.reindex(['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'])
day_revenue.plot(kind='bar', title='Revenue by Day of Week')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

### 9. Profitability (Stretch)
Assume profit margin = 20% of revenue. Estimate total profit and identify the most profitable region.

```python
# Your code here
# Assume profit margin is 20% of revenue
data['profit'] = data['revenue'] * 0.20

# Calculate total profit
total_profit = data['profit'].sum()
print("Estimated total profit:", total_profit)

# Most profitable region
region_profits = data.groupby('region')['profit'].sum().sort_values(ascending=False)
top_profit_region = region_profits.idxmax()
print("Region with highest total profit:", top_profit_region)
print("Profits by region:")
print(region_profits)
```
