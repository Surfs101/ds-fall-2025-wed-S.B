## Step 1: Data Loading and Exploration

Start here! Load the dataset and get familiar with what you're working with.


```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("/workspaces/ds-fall-2025-wed-S.B/Week-02-Pandas-Part-2-and-DS-Overview/data/Ask A Manager Salary Survey 2021 (Responses) - Form Responses 1.tsv", sep='\t')


print(data.head())
print(data.info())
print(data.describe())
print(data.isnull().sum())
```

## Step 2: Data Cleaning


```python
# Convert date columns to datetime
data['Timestamp'] = pd.to_datetime(data['Timestamp'])

# Remove duplicates
data = data.drop_duplicates()

# Fill or drop missing values (if any)
# Example: data['column_name'] = data['column_name'].fillna(fill_value)
# Or: data = data.dropna()

# Check datatypes and fix where needed
# Example: data['quantity'] = data['quantity'].astype(int)

# Confirm cleaning complete
print(data.info())

```

## Step 3: Business Questions Analysis

Now answer those important business questions!


```python
# Question 1: What is the median salary for Software Engineers in the United States?


column_name = "What is your annual salary? (You'll indicate the currency in a later question. If you are part-time or hourly, please enter an annualized equivalent -- what you would earn if you worked the job 40 hours a week, 52 weeks a year.)"

#Filter for US-based Software Engineers (update column names if needed)
us_se = data[
    (data['Job title'].str.contains('Software Engineer', case=False, na=False)) &
    (data['What country do you work in?'] == 'United States')
]

median_salary = us_se[column_name].median()
print("Median US Software Engineer salary:", median_salary)

```

```python
# Question 2: Which US state has the highest average salary for tech workers?
# Group by State and get mean salary (assuming there's a 'State' column)
state_avg_salary = data.groupby('State')["What is your annual salary? (You'll indicate the currency in a later question. If you are part-time or hourly, please enter an annualized equivalent -- what you would earn if you worked the job 40 hours a week, 52 weeks a year.)"].mean()
top_state = state_avg_salary.idxmax()
print("State with highest average salary:", top_state)

```

```python
# Question 3: How much does salary increase on average for each year of experience in tech?
# Use regression for more robust analysis if possible
import numpy as np
exp = data['Years of Experience'].replace('n/a', np.nan).astype(float)
salary = data["What is your annual salary? (You'll indicate the currency in a later question. If you are part-time or hourly, please enter an annualized equivalent -- what you would earn if you worked the job 40 hours a week, 52 weeks a year.)"]
valid = exp.notnull() & salary.notnull()
coeff = np.polyfit(exp[valid], salary[valid], 1)[0]
print("Estimated salary increase per year:", coeff)
```

```python

# Question 4: What percentage of respondents work remotely vs. in-office?
# Calculate remote vs in-office
remote = data['Remote'].str.contains('remote', case=False, na=False)
remote_pct = remote.mean() * 100
in_office_pct = (~remote).mean() * 100
print(f"Remote: {remote_pct:.2f}%, In-office: {in_office_pct:.2f}%")
```

```python
# Question 5: Which industry (besides tech) has the highest median salary?
# Group by industry and compute median salary
industry_medians = data.groupby('What industry do you work in?')['What is your annual salary?'].median()
non_tech = industry_medians.drop(['Tech', 'Technology'], errors='ignore')
top_industry = non_tech.idxmax()
print("Non-tech industry with highest median salary:", top_industry)
```
